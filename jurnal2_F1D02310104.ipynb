```python
import numpy as np
import cv2 as cv
import matplotlib.pyplot as plt
```

## gambar asli

```python
gambar_asli = plt.imread('Assets/aktivis.jpg')
plt.imshow(gambar_asli, cmap='gray')
plt.axis('off')
```

## urutan gambar

```python
gambar_asli = cv.cvtColor(gambar_asli, cv.COLOR_BGR2GRAY)

gambar1 = gambar_asli[0:1300, 0:455]
gambar2 = gambar_asli[0:1300, 456:908]
gambar3 = gambar_asli[0:1300, 909:1360]
gambar4 = gambar_asli[0:1300, 1361:1815]
gambar5 = gambar_asli[0:1300, 1816:2270]
gambar6 = gambar_asli[0:1300, 2271:2720]

tinggi = gambar1.shape[0]
lebar1 = gambar1.shape[1]
lebar2 = gambar2.shape[1]
lebar3 = gambar3.shape[1]
lebar4 = gambar4.shape[1]
lebar5 = gambar5.shape[1]
lebar6 = gambar6.shape[1]

total_lebar = lebar2 + lebar6 + lebar4 + lebar1 + lebar5 + lebar3

gambar_baru = np.zeros((tinggi, total_lebar), dtype=gambar1.dtype)

current_x = 0
gambar_baru[:, current_x:current_x + lebar2] = gambar2
current_x += lebar2

gambar_baru[:, current_x:current_x + lebar6] = gambar6
current_x += lebar6

gambar_baru[:, current_x:current_x + lebar4] = gambar4
current_x += lebar4

gambar_baru[:, current_x:current_x + lebar1] = gambar1
current_x += lebar1

gambar_baru[:, current_x:current_x + lebar5] = gambar5
current_x += lebar5

gambar_baru[:, current_x:current_x + lebar3] = gambar3

plt.imshow(gambar_baru, cmap='gray')
plt.axis('off')
plt.show()
```

## ekualisasi

```python
if gambar_baru.max() <= 1.0:
    gambar_baru = (gambar_baru * 255).astype(np.uint8)

if len(gambar_baru.shape) == 3:
    gambar_baru = cv.cvtColor(gambar_baru, cv.COLOR_RGB2GRAY)
```
